**2. Очищення даних**

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# Видаляти рядки з пропущеними значеннями 
df_cleaned = df.dropna()
df_cleaned
```

```python
# Видалити колонки з пропущеними значеннями більше ніж 30%

df_cleaned_1 = df.dropna(axis=1, thresh=len(df)*0.7)
df_cleaned_1
```

```python
df_cleaned_1.age.hist()
```

```python
df_cleaned_1['age_non_filled'] = df_cleaned_1['age'].isna()
df_cleaned_1.columns
```

```python
# Заповнення пропущених значень у колонці embarked модою (найчастіше зустрічається значення)
df_cleaned_1.fillna({'age': df_cleaned_1['age'].median()}, inplace=True)
```

```python
# Заповнення пропущених значень у колонці embarked модою (найчастіше зустрічається значення) через переприсвоєння нового значення колонки
# df_cleaned_1.embarked.unique()

df_cleaned_1['embarked'] = df_cleaned_1['embarked'].fillna(df_cleaned_1['embarked'].mode()[0])
```

```python
df_cleaned_1.isna().sum()
```

```python
# Видалити дублікати рядків з однаковими значеннями у колонках name та age
df_cleaned_1.drop_duplicates(subset=['name', 'age'], inplace=True)
```

```python
# Перейменувати колонки
df_cleaned_1.rename(columns={'sex': 'gender', 'age': 'passenger_age'}, inplace=True)
```

```python
df_cleaned_1
```

```python
df = df_cleaned_1
```

```python
df
```

```python
df.rename(columns={'gender': 'sex'}, inplace=True)
df
```

```python
# Створити нову колонку
df['age_group'] = pd.cut(df['age'], 
                         bins=[0, 18, 35, 60, 100], 
                         labels=['child', 'young', 'middle', 'senior'])
```

```python
df.age_group.value_counts()
```

```python
df['is_child'] = (df['age'] < 18).astype(int)
```

```python
df.is_child.value_counts()
```

```python
# Розмір сім'ї
df['family_size'] = df['sibsp'] + df['parch'] + 1
df.family_size.value_counts()
```

```python
df['is_alone'] = (df['family_size'] == 1).astype(int)
df.is_alone.value_counts()
```

```python
# Витягнути титул з імені
df['title'] = df['name'].str.extract(r',\s*([A-Za-z]+)\.', expand=False)
df.title.value_counts()
```

```python
# Категорізація fare
df['fare_category'] = pd.qcut(df['fare'], q=4, labels=['low', 'medium', 'high', 'very_high'])
df.fare_category.value_counts()
```

```python
# Взаємодія колонок
df['age_class'] = df['age'] * df['pclass']
df.age_class.value_counts()
```

**3. Групування та агрегація**


```python
# Групування за однією колонкою
by_class = df.groupby('pclass')['survived'].mean()
print("Виживання за класом:")
by_class
```

```python
# Групування за кількома колонками
by_class_sex = df.groupby(['pclass', 'sex'])['survived'].mean()
by_class_sex
```

```python
type(by_class_sex)
```

```python
# Кілька агрегацій одночасно
statistics = df.groupby('pclass').agg({
    'age': ['mean', 'median', 'std'],
    'fare': ['mean', 'max', 'min'],
    'survived': ['sum', 'count', 'mean']
})

statistics.round(2)
```

```python
# Custom агрегація 
def range_func(x):
    return x.max() - x.min()

df.groupby('pclass')['age'].agg(['mean', 'median', range_func]).round(2)
```

```python
# Використання size, бо з агрегацією value_counts() повертає помилку
df.groupby(['pclass', 'survived']).size()
```
